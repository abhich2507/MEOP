```python
import numpy as np
import matplotlib.pyplot as plt

# --- 1. NUMERICAL EXAMPLE ---
# We create a simple array with a sudden noise spike (the 20.0)
deriv = np.array([2.0, 4.0, 6.0, 20.0, 8.0, 10.0, 12.0])

# Define the moving average window size
window = 3

# Create the kernel and apply smoothing
kernel = np.ones(window) / window
deriv_smooth = np.convolve(deriv, kernel, mode='same')

print("Original array: ", np.round(deriv, 2))
print("Kernel:         ", np.round(kernel, 2))
print("Smoothed array: ", np.round(deriv_smooth, 2))

# --- 2. VISUAL EXAMPLE ---
# Let's create a larger, noisy dataset (like real experimental data)
x = np.linspace(0, 10, 150)
noisy_signal = np.sin(x) + np.random.normal(0, 0.2, 150)

# Smooth it using a window of 10
new_window = 20
smooth_signal = np.convolve(noisy_signal, np.ones(new_window)/new_window, mode='same')

# Plot the results
plt.figure(figsize=(10, 5))
plt.plot(x, noisy_signal, label='Noisy Data (deriv)', alpha=0.5, color='gray', marker='.', linestyle='none')
plt.plot(x, smooth_signal, label=f'Smoothed Data (window={new_window})', color='red', linewidth=2.5)
plt.title('Visualizing Moving Average via np.convolve', fontsize=14)
plt.xlabel('Time / Index')
plt.ylabel('Amplitude')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.savefig('smoothing_plot.png')
```
